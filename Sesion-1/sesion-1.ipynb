## 🧩 5. Funciones de orden superior (`map`, `filter`, `reduce`)

Son funciones que **reciben otras funciones** o **devuelven funciones**.


### 🔹 `map`

Aplica una función a cada elemento de un iterable.


### 🔹 `filter`

Filtra elementos según una condición.


### 🔹 `reduce`

Acumula los valores de una lista según una función.

> Necesita importarse desde `functools`.


### 🧩 Actividad 11

   - La lista original  
   - La lista de cuadrados (con `map`)
