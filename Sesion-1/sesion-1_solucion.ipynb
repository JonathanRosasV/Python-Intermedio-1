```python
# 1. Crear una lista con los números del 1 al 30.
# range(1, 31) genera los números del 1 al 30 (el final no se incluye).
# Lo convertimos a lista para poder verla completa.
numeros = list(range(1, 31))

# 2. Lista con solo los múltiplos de 3.
# Uso comprensión de listas porque es más claro y muy usado en Python.
# La condición n % 3 == 0 significa "el residuo es cero", es decir, múltiplo.
multiplos_3 = [n for n in numeros if n % 3 == 0]

# 3. Lista con los cuadrados de los números pares.
# Nuevamente usamos comprensión de listas, pero esta vez comprobamos si son pares
# y luego elevamos al cuadrado con n**2.
cuadrados_pares = [n**2 for n in numeros if n % 2 == 0]

# Imprimimos los resultados con mensajes claros.
print("Lista de 1 a 30:", numeros)
print("Múltiplos de 3:", multiplos_3)
print("Cuadrados de los números pares:", cuadrados_pares)
```

### 🔹 Challenge 1: FizzBuzz

> De 1 a 100:  
> - Si el número es múltiplo de 3 → `"Fizz"`  
> - Si es múltiplo de 5 → `"Buzz"`  
> - Si es múltiplo de 3 y 5 → `"FizzBuzz"`  
> - Si no, imprime el número.  


```python
# Recorremos los números del 1 al 100.
# range(1, 101) llega hasta el 100 (el límite no se incluye).
for n in range(1, 101):

    # Primero verificamos el caso más restrictivo:
    # múltiplo de 3 y de 5 al mismo tiempo (o sea, múltiplo de 15).
    # Si no ponemos esta condición primero, los otros casos se disparan antes.
    if n % 3 == 0 and n % 5 == 0:
        print("FizzBuzz")

    # Si no cae en el caso anterior, revisamos si es múltiplo de 3.
    elif n % 3 == 0:
        print("Fizz")

    # Luego revisamos si es múltiplo de 5.
    elif n % 5 == 0:
        print("Buzz")

    # Si no entra en ninguno de los casos anteriores, imprimimos el número tal cual.
    else:
        print(n)
```

### 🔹 Challenge 2: ¿Es un anagrama?

Un anagrama es una palabra formada reordenando **todas** las letras de otra palabra.  
Ejemplo: `"roma"` y `"amor"`.


### 🧩 Actividad 6  

Mejora la función para que **ignore espacios**.  
Ejemplo: `"r om a"` y `"a mor"` también deberían ser anagramas.


```python
def es_anagrama(palabra1, palabra2):
    # Primero eliminamos todos los espacios.
    # replace(" ", "") quita los espacios sin complicarse con expresiones regulares.
    # Es una buena práctica “limpiar” las entradas antes de compararlas.
    p1 = palabra1.replace(" ", "").lower()
    p2 = palabra2.replace(" ", "").lower()

    # Convertimos cada palabra en una lista de letras ordenadas.
    # Si ambas listas son iguales, entonces las palabras son anagramas.
    return sorted(p1) == sorted(p2)


# Probando la función
print(es_anagrama("roma", "amor"))        # True
print(es_anagrama("r om a", "a mor"))     # True
print(es_anagrama("python", "typhon"))    # True
print(es_anagrama("hola", "halo"))        # True
print(es_anagrama("hola", "ola"))         # False
```

### 🔹 Challenge 3: Sucesión de Fibonacci

Imprime los 50 primeros números de la sucesión de Fibonacci.


```python
# Vamos a imprimir los primeros 50 números de la sucesión de Fibonacci.
# Recordemos que comienza con 0 y 1, y cada número es la suma de los dos anteriores.

fib = [0, 1]  # Punto de partida clásico

# Generamos números hasta llegar a 50 elementos.
while len(fib) < 50:
    # El nuevo número es la suma de los últimos dos.
    nuevo = fib[-1] + fib[-2]
    fib.append(nuevo)

# Imprimimos toda la lista.
print(fib)
```

### 🧩 Actividad 7  

Escribe una versión de Fibonacci que reciba un `n` cualquiera y **devuelva** una lista con los primeros `n` términos.


```python
def fibonacci(n):
    # Si piden 0 términos, devolvemos lista vacía
    if n == 0:
        return []
    # Si piden 1 término, devolvemos solo [0]
    if n == 1:
        return [0]

    fib = [0, 1]  # punto de partida

    # Generar hasta tener n elementos
    while len(fib) < n:
        nuevo = fib[-1] + fib[-2]
        fib.append(nuevo)

    return fib

```

### 🔹 Challenge 4: Números primos

Un número primo es divisible solo por 1 y por sí mismo.


### 🧩 Actividad 8  

Crea una función `primos_hasta(n)` que devuelva una lista con todos los números primos entre 2 y `n`.


```python
def primos_hasta(n):
    # Lista donde iremos guardando los números primos encontrados.
    primos = []

    # Recorremos desde 2 hasta n, porque el 1 no se considera primo.
    for num in range(2, n + 1):

        # Suponemos que el número es primo hasta que encontremos un divisor.
        es_primo = True

        # Probamos divisores desde 2 hasta la raíz cuadrada del número.
        # Este truco hace el código más rápido que revisar hasta num-1.
        for divisor in range(2, int(num**0.5) + 1):
            if num % divisor == 0:
                es_primo = False
                break  # No hace falta seguir probando divisores.

        # Si nadie lo dividió, entonces es primo.
        if es_primo:
            primos.append(num)

    return primos


# Ejemplo de uso:
print(primos_hasta(50))
```

Escribe una versión de `reverse` **sin** usar slicing (`[::-1]`), solo con bucles.


```python
def reverse(texto):
    # Vamos a construir la versión invertida letra por letra.
    invertido = ""

    # Recorremos el texto desde el final hasta el inicio.
    # range(len(texto)-1, -1, -1) significa:
    #   - empezar en el último índice
    #   - terminar en el índice -1 (sin incluirlo)
    #   - ir de uno en uno hacia atrás
    for i in range(len(texto) - 1, -1, -1):
        invertido += texto[i]

    return invertido


# Probando la función con UPAO
print(reverse("UPAO"))
```

## 🧩 5. Funciones de orden superior (`map`, `filter`, `reduce`)

Son funciones que **reciben otras funciones** o **devuelven funciones**.


### 🔹 `map`

Aplica una función a cada elemento de un iterable.


```python
def cuadrado(n):
    return n**2

numeros = [1, 2, 3, 4, 5]

resultado = map(cuadrado, numeros)

print(list(resultado))  # [1, 4, 9, 16, 25]
```

### 🔹 `filter`

Filtra elementos según una condición.


```python
def es_par(n):
    return n % 2 == 0

numeros = [1, 2, 3, 4, 5, 6]

resultado = filter(es_par, numeros)

print(list(resultado))  # [2, 4, 6]
```

### 🔹 `reduce`

Acumula los valores de una lista según una función.

> Necesita importarse desde `functools`.


```python
from functools import reduce

numeros = [1, 2, 3, 4, 5]

suma_total = reduce(lambda a, b: a + b, numeros)

print(suma_total)  # 15
```

   - La lista original  
   - La lista de cuadrados (con `map`)  
   - La lista de pares (con `filter`)  


```python
from functools import reduce

# 1. Usar filter para quedarnos con números >= 10.
numeros = [3, 7, 10, 12, 5, 20, 1, 15]

mayores_10 = list(filter(lambda x: x >= 10, numeros))
print("Números >= 10:", mayores_10)


# 2. Usar reduce para obtener el producto de todos los elementos de una lista.
producto_total = reduce(lambda a, b: a * b, numeros)
print("Producto total:", producto_total)


# 3. Función que retorna:
#    - la lista original
#    - la lista de cuadrados (con map)
#    - la lista de pares (con filter)

def analizar_lista(lista):
    # Lista de cuadrados
    cuadrados = list(map(lambda x: x**2, lista))

    # Lista solo con pares
    pares = list(filter(lambda x: x % 2 == 0, lista))

    return lista, cuadrados, pares


# Probamos la función
original, cuadrados, pares = analizar_lista(numeros)

print("Original:", original)
print("Cuadrados:", cuadrados)
print("Pares:", pares)
```
