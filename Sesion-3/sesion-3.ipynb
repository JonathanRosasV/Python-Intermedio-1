## 1. Manejo de fechas e índices temporales

En `pandas`, las fechas se manejan con el tipo `datetime64`. Ideas clave:

- `pd.to_datetime()` convierte texto a fechas.
- Un índice temporal permite usar `resample`, filtros por rango de fechas, etc.
- Podemos extraer año, mes, día de la semana, etc. desde el índice.


```python
import pandas as pd

# Datos diarios de ejemplo
ventas = {
    "fecha": [
        "2024-01-01", "2024-01-02", "2024-01-03", "2024-01-04", "2024-01-05",
        "2024-01-06", "2024-01-07", "2024-01-08", "2024-01-09", "2024-01-10"
    ],
    "monto": [120, 150, 130, 160, 170, 165, 180, 175, 190, 200]
}

df = pd.DataFrame(ventas)

# Conversión a datetime y uso como índice
df['fecha'] = pd.to_datetime(df['fecha'])
df = df.set_index('fecha')

# Extra: descomposición de la fecha
df['anio'] = df.index.year
df['mes'] = df.index.month
try:
    df['dia_semana'] = df.index.day_name(locale='es_ES')
except TypeError:
    # Fallback por si locale no está disponible
    df['dia_semana'] = df.index.day_name()

df
```

## 2. Cambio de frecuencia y agregación (resample, rolling)

- `resample('W').sum()` permite pasar de datos diarios a semanales (u otra frecuencia).
- `rolling(window)` sirve para calcular medias móviles y suavizar la serie.


```python
# Agregación semanal de las ventas (suma)
df_semanal = df['monto'].resample('W').sum()
print("Ventas semanales (suma):")
print(df_semanal)

# Media móvil de 3 días
df['monto_mm3'] = df['monto'].rolling(3).mean()
print("\nDataFrame con media móvil de 3 días:")
print(df)
```

## 3. Tasas de variación y acumulación temporal

La variación porcentual entre dos periodos consecutivos se calcula como:

$$\text{Var\%}_t = \frac{x_t - x_{t-1}}{x_{t-1}} \times 100$$

El acumulado se obtiene con `cumsum()`.


```python
# Variación porcentual diaria
df['var_pct'] = df['monto'].pct_change() * 100

# Monto acumulado
df['acumulado'] = df['monto'].cumsum()

df
```

## 4. Visualización con Matplotlib y Seaborn

Graficar series de tiempo permite identificar rápidamente tendencias, picos y caídas.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Gráfico con Matplotlib
plt.figure(figsize=(8, 4))
plt.plot(df.index, df['monto'])
plt.title('Ventas diarias')
plt.xlabel('Fecha')
plt.ylabel('Monto (S/)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# Gráfico con Seaborn
plt.figure(figsize=(8, 4))
sns.lineplot(x=df.index, y=df['monto'])
plt.title('Ventas diarias (Seaborn)')
plt.xlabel('Fecha')
plt.ylabel('Monto (S/)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Personalización de gráficos

Podemos cambiar estilo global, colores, grosor de líneas, añadir rejilla, etc., para hacer los gráficos más claros.


```python
plt.style.use('seaborn-v0_8-darkgrid')

plt.figure(figsize=(8, 4))
plt.plot(df.index, df['monto'], color='purple', linewidth=2, marker='o')
plt.title('Ventas diarias (gráfico personalizado)')
plt.xlabel('Fecha')
plt.ylabel('Monto (S/)')
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

## 6. Visualizaciones comparativas y tendencias

Comparar la serie original con una media móvil permite separar ruido de corto plazo y tendencia.


```python
plt.figure(figsize=(9, 4))

# Serie original
plt.plot(df.index, df['monto'], label='Ventas diarias', marker='o')

# Media móvil 3 días
plt.plot(df.index, df['monto_mm3'], label='Media móvil 3 días', linewidth=3)

plt.title('Ventas diarias vs. media móvil de 3 días')
plt.xlabel('Fecha')
plt.ylabel('Monto (S/)')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
