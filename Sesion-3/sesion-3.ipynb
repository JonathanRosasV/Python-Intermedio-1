## 0. Cargar base de datos

La siguiente celda **simula** la carga de una base de ventas diarias llamada `df`.
En un contexto real podrías usar `pd.read_csv('ventas_diarias.csv')` u otra fuente.


```python
import pandas as pd
import numpy as np

# Simulamos una base de datos de 60 días de ventas diarias
np.random.seed(42)
fechas = pd.date_range(start='2024-01-01', periods=60, freq='D')
tendencia = np.linspace(100, 160, 60)  # tendencia creciente suave
ruido = np.random.randint(-10, 11, size=60)  # ruido entre -10 y 10
monto = tendencia + ruido

df = pd.DataFrame({
    'fecha': fechas,
    'monto': np.round(monto, 2)
})

df.head()
```

## 1. Manejo de fechas e índices temporales

En `pandas`, las fechas se manejan con el tipo `datetime64`. Ideas clave:

- `pd.to_datetime()` convierte texto a fechas.
- Un índice temporal permite usar `resample`, filtros por rango de fechas, etc.
- Podemos extraer año, mes, día de la semana, etc. desde el índice.


```python
# 📝 ACTIVIDAD 1: Manejo de fechas e índices temporales
# Objetivo:
# 1) Asegurar que la columna 'fecha' esté en formato datetime.
# 2) Usar 'fecha' como índice del DataFrame.
# 3) Crear las columnas 'anio', 'mes' y 'dia_semana'.

# Escribe tu solución aquí:
# df['fecha'] = ...
# df = ...
# df['anio'] = ...
# df['mes'] = ...
# df['dia_semana'] = ...

# df.head()
```

```python
# ✅ SOLUCIÓN PROPUESTA ACTIVIDAD 1
df['fecha'] = pd.to_datetime(df['fecha'])
df = df.set_index('fecha')

df['anio'] = df.index.year
df['mes'] = df.index.month
try:
    df['dia_semana'] = df.index.day_name(locale='es_ES')
except TypeError:
    df['dia_semana'] = df.index.day_name()

df.head()
```

## 2. Cambio de frecuencia y agregación (resample, rolling)

- `resample('W').sum()` permite pasar de datos diarios a semanales (u otra frecuencia).
- `rolling(window)` sirve para calcular medias móviles y suavizar la serie.


```python
# 📝 ACTIVIDAD 2: Resample y medias móviles
# Objetivos:
# 1) Crear una serie de ventas semanales (suma) a partir de df['monto'].
# 2) Crear una media móvil de 7 días de la columna 'monto'.

# Escribe tu solución aquí:
# df_semanal = ...
# df['monto_mm7'] = ...

# df_semanal.head()
# df[['monto', 'monto_mm7']].head(10)
```

```python
# ✅ SOLUCIÓN PROPUESTA ACTIVIDAD 2
df_semanal = df['monto'].resample('W').sum()
df['monto_mm7'] = df['monto'].rolling(7).mean()

print("Ventas semanales (suma):")
print(df_semanal.head())

print("\nVista de la serie con media móvil de 7 días:")
df[['monto', 'monto_mm7']].head(15)
```

## 3. Tasas de variación y acumulación temporal

La variación porcentual entre dos periodos consecutivos se calcula como:

$$\text{Var\%}_t = \frac{x_t - x_{t-1}}{x_{t-1}} \times 100$$

El acumulado se obtiene con `cumsum()`.


```python
# 📝 ACTIVIDAD 3: Variación porcentual y acumulado
# Objetivos:
# 1) Crear la columna 'var_pct' con la variación porcentual diaria de 'monto'.
# 2) Crear la columna 'acumulado' con el monto acumulado en el tiempo.

# Escribe tu solución aquí:
# df['var_pct'] = ...
# df['acumulado'] = ...

# df[['monto', 'var_pct', 'acumulado']].head(10)
```

```python
# ✅ SOLUCIÓN PROPUESTA ACTIVIDAD 3
df['var_pct'] = df['monto'].pct_change() * 100
df['acumulado'] = df['monto'].cumsum()

df[['monto', 'var_pct', 'acumulado']].head(10)
```

## 4. Visualización con Matplotlib y Seaborn

Graficar series de tiempo permite identificar rápidamente tendencias, picos y caídas.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# 📝 ACTIVIDAD 4: Gráfico de la serie de ventas
# Objetivos:
# 1) Graficar 'monto' en el tiempo usando Matplotlib.
# 2) Repetir el gráfico usando Seaborn.

# Escribe tu solución aquí:
# plt.figure(...)
# plt.plot(...)
# ...

# plt.figure(...)
# sns.lineplot(...)
```

```python
# ✅ SOLUCIÓN PROPUESTA ACTIVIDAD 4
plt.figure(figsize=(8, 4))
plt.plot(df.index, df['monto'])
plt.title('Ventas diarias')
plt.xlabel('Fecha')
plt.ylabel('Monto (S/)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
sns.lineplot(x=df.index, y=df['monto'])
plt.title('Ventas diarias (Seaborn)')
plt.xlabel('Fecha')
plt.ylabel('Monto (S/)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Visualizaciones comparativas y tendencias

Comparar la serie original con una media móvil permite separar ruido de corto plazo y tendencia.


```python
# 📝 ACTIVIDAD 6: Serie original vs media móvil
# Objetivos:
# 1) Graficar en el mismo gráfico la serie original 'monto'.
# 2) Añadir la media móvil de 7 días 'monto_mm7' en la misma figura.
# 3) Incluir leyenda, título y etiquetas.

# Escribe tu solución aquí:
# plt.figure(...)
# plt.plot(..., label='Ventas diarias')
# plt.plot(..., label='Media móvil 7 días')
# plt.legend()
# plt.title(...)
# plt.xlabel(...)
# plt.ylabel(...)
# plt.show()
```
